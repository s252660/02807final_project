# LDA topic mixtures: first 10 songs per k

Prints the first 10 rows of `song_topics.parquet` for each run in `data/lda_grid_search/`.

```python
import re
from pathlib import Path

import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.precision", 3)

ROOT = Path.cwd()
if not (ROOT / "data").exists():  # running from notebooks/
    ROOT = ROOT.parent
GRID_DIR = ROOT / "data" / "lda_grid_search"

runs = {}
for d in GRID_DIR.iterdir():
    m = re.match(r"k(\d+)_seed(\d+)", d.name)
    if m and (d / "song_topics.parquet").exists():
        runs[int(m.group(1))] = d
runs = dict(sorted(runs.items()))
print("k values found:", list(runs))
```

```python
for k, d in runs.items():
    print(f"k = {k}")
    display(pd.read_parquet(d / "song_topics.parquet").set_index("song").head(10))
```
